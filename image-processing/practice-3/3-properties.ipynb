```python
import cv2 as cv
image_path = "image.png"
img = cv.imread(image_path, cv.IMREAD_GRAYSCALE)
```

```python
import numpy as np
def average(image):
    flattened_image = np.array(image).flatten()
    return np.average(flattened_image)
def variance(image):
    flattened_image = np.array(image).flatten()
    return np.var(flattened_image)
def gravity(image):
    height, width = image.shape
    y_indices, x_indices = np.indices((height, width))
    total_pixels = np.sum(image)
    total_x = np.sum(x_indices * image)
    total_y = np.sum(y_indices * image)
    return (float(total_x / total_pixels), float(total_y / total_pixels))
```

```python
print(average(img))
print(variance(img))
print(gravity(img))
```
