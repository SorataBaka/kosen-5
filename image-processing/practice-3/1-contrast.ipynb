```python
image_path = "image.png"
min_output = 10
max_output = 240
```

```python
import cv2 as cv
img = cv.imread(image_path, cv.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError(f"Image not found at path: {image_path}")
img = img.astype(float)
height, width = img.shape
print(f"Width: {width}, Height: {height}")
```

```python
# find min and max pixel values in the image
if img is None:
    raise ValueError("Image data is not loaded properly.")

min_pixel = img.min()
max_pixel = img.max()
print(f"Min pixel value: {min_pixel}, Max pixel value: {max_pixel}")
        
```

```python
for h in range(height):
    for w in range(width):
        pixel_value = img[h, w]
        # apply contrast stretching formula
        new_value = (((max_output - min_output) * (pixel_value - min_pixel)) / (max_pixel - min_pixel)) + min_output
        # clip the value to be in valid range [0, 255]
        img[h, w] = new_value
```

```python
import matplotlib.pyplot as plt
plt.imshow(img)
plt.axis("off")
plt.show()
```
