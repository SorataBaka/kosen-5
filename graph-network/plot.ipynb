```python
import matplotlib.pyplot as plt
import numpy as np

# データ
files = [
    "search_100s.dat",
    "search_100d.dat",
    "search_500s.dat",
    "search_500d.dat",
    "search_1000s.dat",
    "search_1000d.dat"
]
dfs_depth = [23, 5, 37, 6, 48, 9]
bfs_depth = [5, 3, 4, 3, 3, 3]

# X 軸の位置
x = np.arange(len(files))
width = 0.35  # 棒の幅

# グラフ描画
fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, dfs_depth, width, label="DFSの深さ")
rects2 = ax.bar(x + width/2, bfs_depth, width, label="BFSの深さ")

plt.rcParams['font.family'] = 'Hiragino Sans'
# ラベル・タイトル
ax.set_xlabel("ファイル名")
ax.set_ylabel("探索の深さ")
ax.set_title("DFSとBFSの深さの比較（ファイル別）")
ax.set_xticks(x)
ax.set_xticklabels(files, rotation=45, ha="right")
ax.legend()

# 棒の上に数値を表示
for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f"{height}",
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3),  # 少し上にずらす
                textcoords="offset points",
                ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# データ
files = [
    "search_100s.dat",
    "search_100d.dat",
    "search_500s.dat",
    "search_500d.dat",
    "search_1000s.dat",
    "search_1000d.dat"
]
dfs_leaf = [73, 96, 457, 495, 947, 992]
bfs_leaf = [70, 94, 453, 493, 946, 993]
diff_percent = [4.29, 2.13, 0.88, 0.41, 0.11, -0.10]

x = np.arange(len(files))
width = 0.35

# --- 棒グラフ（DFSとBFSの葉ノード数の比較） ---
fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, dfs_leaf, width, label="DFSの葉ノード数")
rects2 = ax.bar(x + width/2, bfs_leaf, width, label="BFSの葉ノード数")

ax.set_xlabel("ファイル名")
ax.set_ylabel("葉ノード数")
ax.set_title("DFSとBFSにおける全域木の葉ノード数の比較")
ax.set_xticks(x)
ax.set_xticklabels(files, rotation=45, ha="right")
ax.legend()

# 棒の上に値を表示
for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f"{height}",
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3),
                textcoords="offset points",
                ha="center", va="bottom")

plt.tight_layout()
plt.show()

# --- 差分（％）の棒グラフ ---
fig, ax = plt.subplots(figsize=(10, 6))
rects = ax.bar(files, diff_percent, color="orange")

ax.set_xlabel("ファイル名")
ax.set_ylabel("差（%）")
ax.set_title("DFSとBFSの葉ノード数の差（%）")
ax.set_xticklabels(files, rotation=45, ha="right")

# 値を表示
for rect, value in zip(rects, diff_percent):
    height = rect.get_height()
    ax.annotate(f"{value:.2f}%",
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3 if value >= 0 else -12),
                textcoords="offset points",
                ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# データ
files = [
    "search_100s.dat",
    "search_100d.dat",
    "search_500s.dat",
    "search_500d.dat",
    "search_1000s.dat",
    "search_1000d.dat"
]
bfs_diameter = [7, 4, 5, 4, 4, 4]
dfs_diameter = [23, 5, 37, 6, 48, 9]
change_percent = [229, 25, 640, 50, 1100, 125]  # +は省略して数値だけ保持

x = np.arange(len(files))
width = 0.35

# --- 棒グラフ（BFSとDFSの直径の比較） ---
fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, bfs_diameter, width, label="BFS直径")
rects2 = ax.bar(x + width/2, dfs_diameter, width, label="DFS直径")

ax.set_xlabel("ファイル名")
ax.set_ylabel("直径")
ax.set_title("DFSとBFSにおける全域木の直径の比較")
ax.set_xticks(x)
ax.set_xticklabels(files, rotation=45, ha="right")
ax.legend()

# 棒の上に値を表示
for rect, value in zip(rects1, bfs_diameter):
    ax.annotate(f"{value}",
                xy=(rect.get_x() + rect.get_width()/2, rect.get_height()),
                xytext=(0, 3),
                textcoords="offset points",
                ha="center", va="bottom")

for rect, value in zip(rects2, dfs_diameter):
    ax.annotate(f"{value}",
                xy=(rect.get_x() + rect.get_width()/2, rect.get_height()),
                xytext=(0, 3),
                textcoords="offset points",
                ha="center", va="bottom")

plt.tight_layout()
plt.show()

# --- 差分（％）の棒グラフ ---
fig, ax = plt.subplots(figsize=(10, 6))
rects = ax.bar(files, change_percent, color="orange")

ax.set_xlabel("ファイル名")
ax.set_ylabel("変化率（%）")
ax.set_title("DFSとBFSの直径の変化率（%）")
ax.set_xticklabels(files, rotation=45, ha="right")

# 値を表示
for rect, value in zip(rects, change_percent):
    height = rect.get_height()
    ax.annotate(f"+{value}%",
                xy=(rect.get_x() + rect.get_width()/2, height),
                xytext=(0, 3),
                textcoords="offset points",
                ha="center", va="bottom")

plt.tight_layout()
plt.show()
```
