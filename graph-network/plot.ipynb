```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'Hiragino Sans'
n = list(range(1, 300))        # n from 1 to 20
y = [2**i for i in n]         # y = n^2

plt.plot(n, y, marker='o')
plt.title("最大クリーク探索における探索空間の成長")
plt.xlabel("ノード数")
plt.ylabel("調べる組み合わせの数")
plt.grid(True)
plt.show()
```
